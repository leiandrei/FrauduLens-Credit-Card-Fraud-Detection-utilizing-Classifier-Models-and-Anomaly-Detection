## Exploratory Data Analysis: Initial Dataset Understanding

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

sns.set_style('whitegrid')
df = pd.read_csv('../../data/raw/creditcard.csv')
df.head()
```

### Data Inspection

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.shape
```

```python
df.describe()
```

```python
df.duplicated().sum()
df = df.drop_duplicates()
```

```python
# optional: outlier check
numerical_cols = df.drop('Class', axis=1)          
numerical_cols = numerical_cols.columns.to_list()

df_filtered = df.copy()

lower_bound = df[numerical_cols].quantile(0.25)
upper_bound = df[numerical_cols].quantile(0.75)

iqr = upper_bound - lower_bound

lower = lower_bound - 1.5 * iqr
upper = upper_bound + 1.5 * iqr

df_filtered = df_filtered[
    ((df_filtered[numerical_cols] >= lower) &
    (df_filtered[numerical_cols] <= upper)).any(axis=1)]

print(f"Rows truncated after outliers: {df.shape[0] - df_filtered.shape[0]}")
```

### Data Visualization

```python
# correlation heatmap
sns.heatmap(df.corr(), annot=False, cmap='magma')
plt.title('Correlation Heatmap')
```

```python
fig, ax = plt.subplots(figsize=(15, 15))
df.hist(ax=ax)
```

```python
# with log scale
fig, ax = plt.subplots(figsize=(15, 15))
df.hist(ax=ax, alpha=0.8, log=True)
```

```python
fig, ax = plt.subplots(figsize=(8, 6))

class_counts = df['Class'].value_counts()
class_counts.plot(kind='bar', ax=ax, alpha=0.8, edgecolor='black')
plt.xlabel('Class')
plt.ylabel('Count')
plt.title('Class Counts')
plt.tight_layout()
sns.despine(top=True, right=True)
plt.bar_label(ax.containers[0])


print(class_counts)
```

### Feature Significance

```python
feats = df.corr()['Class']

fig, ax = plt.subplots(figsize=(10, 10))
color_lbl = ['red' if val <= 0 else 'green' for val in feats]
feats.plot(kind='barh', alpha=0.8, ax=ax, color=color_lbl)
ax.set_ylabel('Features')
ax.set_xlabel('Correlation')
ax.set_title('Feature Significance')
sns.despine(top=True, right=True)
```
